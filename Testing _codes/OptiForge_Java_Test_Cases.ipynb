## Algorithm

The program reads an integer `n` followed by `n` integers and calculates:

1. Sum of all elements
2. Minimum value
3. Maximum value
4. Number of even elements
5. Longest strictly increasing contiguous run

### Expected complexity

- **Time:** `O(n)`
- **Space:** `O(n)` because the input array is stored

This makes the program useful for testing both correctness and optimization transformations.


## Baseline Source Code

The following code is the reference implementation for this language.

```python
import java.util.Scanner;

public class Main {
    public static void main(String[] args) {
        Scanner scanner = new Scanner(System.in);

        int n = scanner.nextInt();
        long[] values = new long[n];

        for (int i = 0; i < n; i++) {
            values[i] = scanner.nextLong();
        }

        long total = 0;
        long minimum = values[0];
        long maximum = values[0];
        int evenCount = 0;

        int longestIncreasingRun = 1;
        int currentRun = 1;

        for (int i = 0; i < n; i++) {
            long value = values[i];

            total += value;

            if (value < minimum) {
                minimum = value;
            }

            if (value > maximum) {
                maximum = value;
            }

            if (value % 2 == 0) {
                evenCount++;
            }

            if (i > 0) {
                if (values[i] > values[i - 1]) {
                    currentRun++;
                } else {
                    currentRun = 1;
                }

                if (currentRun > longestIncreasingRun) {
                    longestIncreasingRun = currentRun;
                }
            }
        }

        System.out.println("Sum: " + total);
        System.out.println("Minimum: " + minimum);
        System.out.println("Maximum: " + maximum);
        System.out.println("Even count: " + evenCount);
        System.out.println("Longest increasing run: " + longestIncreasingRun);
    }
}
```
