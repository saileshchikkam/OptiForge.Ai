# OptiForge — Cross-Language Verification Test Suite

## Overview

This is the master test specification for validating OptiForge's translation and optimization pipeline across **Python, C++, and Java**.

The baseline implementations in the three language notebooks are intentionally equivalent.

## What this suite tests

- Compilation
- Runtime execution
- stdin parsing
- integer arithmetic
- arrays
- loops
- conditions
- min/max tracking
- even-number counting
- increasing-run detection
- exact stdout comparison
- cross-language translation
- same-language optimization

## Test coverage

There are **8 independent deterministic test cases**.

Each case must be run separately against every source → target transformation.

### Expected behavior

For a correct transformation, the generated program must match the reference output exactly.

## Test matrix

| # | Source | Target |

With 8 test cases, this produces **72 source/target/test combinations per model**.

With 4 coding models, the complete regression suite contains **288 combinations**.
